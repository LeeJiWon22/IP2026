# Morphological Transformations

```python
import cv2
import numpy as np
```

```python
img = cv2.imread('j.png',0)
```

```python
kernel = np.ones((5,5),np.uint8)
erosion = cv2.erode(img,kernel,iterations = 1)
cv2.imshow('this is j',erosion)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dilation = cv2.dilate(img,kernel,iterations = 1)
cv2.imshow('this is j',dilation)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img2 = cv2.imread('j2.png',0)
opening = cv2.morphologyEx(img2, cv2.MORPH_OPEN, kernel)
cv2.imshow('this is j1',img2)
cv2.imshow('this is j2',opening)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img3 = cv2.imread('j3.png',0)
closing = cv2.morphologyEx(img3, cv2.MORPH_CLOSE, kernel)
cv2.imshow('this is j1',img3)
cv2.imshow('this is j2',closing)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
