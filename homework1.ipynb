```python
import numpy as np 
import cv2
```

```python
def nothing(x):
    pass
```

```python
drawing = False # true if mouse is pressed
mode = True # if True, draw rectangle. Press 'm' to toggle to curve
ix,iy = -1,-1
x,y = -1,-1
org_img = cv2.imread('messi5.jpg')

def draw_circle(event,x,y,flags,param):
    global ix,iy,drawing,mode
    
    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True
        ix,iy = x,y
    elif event == cv2.EVENT_MOUSEMOVE:
        if drawing == True:
            if mode == True:
                cv2.rectangle(org_img,(ix,iy),(x,y),(0,255,0),-1)
            else:
                cv2.circle(org_img,(x,y),5,(0,0,255),-1)
    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False
        if mode == True:
            cv2.rectangle(org_img,(ix,iy),(x,y),(0,255,0),-1)
        else:
            cv2.circle(org_img,(x,y),5,(0,0,255),-1)
```

```python
#img = cv2.imread('messi5.jpg') #이미지를 불러옴


cv2.namedWindow('image')
cv2.setMouseCallback('image',draw_circle)
cv2.createTrackbar('R','image',0,255,nothing)

font = cv2.FONT_HERSHEY_SIMPLEX

while(1):
    #cv2.imshow('image',org_img)
    r = cv2.getTrackbarPos('R','image')
    org_img[iy:y,ix:x,2] = [r]
    img = org_img.copy()
    value = cv2.getTrackbarPos('R','image')
    cv2.putText(img,'Mouse position : ',(10,30), font, 0.7,(255,255,255),1,cv2.LINE_AA)
    cv2.imshow('image', img)
    k = cv2.waitKey(1) & 0xFF
    if k == 27:
        break
cv2.destroyAllWindows()
```
