```python
import numpy as np 
import cv2
```

```python
def nothing(x):
    pass
```

```python
drawing = False # true if mouse is pressed
ix,iy = -1,-1
cx,cy = -1, -1
org_img = cv2.imread('messi5.jpg')

def draw_circle(event,x,y,flags,param):
    global ix,iy,cx,cy,drawing

    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True
        ix,iy = x,y

    elif event == cv2.EVENT_MOUSEMOVE:
        cx,cy = x,y

    elif event == cv2.EVENT_LBUTTONUP:
        ix,iy = -1,-1
        cx, cy = -1,-1
        drawing = False
        
```

```python
#img = cv2.imread('messi5.jpg') #이미지를 불러옴


cv2.namedWindow('image')
cv2.setMouseCallback('image',draw_circle)
cv2.createTrackbar('R','image',0,255,nothing)

font = cv2.FONT_HERSHEY_SIMPLEX

while(1):
    img = org_img.copy()

    r = cv2.getTrackbarPos('R','image')
    org_img[iy:cy,ix:cx,2] = [r]

    value = cv2.getTrackbarPos('R','image')
    if drawing == True:
            cv2.rectangle(img,(ix,iy),(cx,cy),(0,255,0),-1)
    cv2.putText(img,f'Mouse position : ({ix},{iy})-({cx},{cy})-{value}',(10,30), font, 0.7,(255,255,255),1,cv2.LINE_AA)
    
    cv2.imshow('image', img)

    k = cv2.waitKey(1) & 0xFF
    if k == 27:
        break
cv2.destroyAllWindows()
```
